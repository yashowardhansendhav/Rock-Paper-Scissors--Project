```python
import random
```

```python
choices = ["rock","paper","scissors"]
```

```python
def determine_winner(player, computer):
    if player == computer: return "draw"

    if player == "rock" and computer == "scissors":
        return "player wins"
    
    if player == "paper" and computer == "rock":
        return "player wins"
    
    if player == "scissors" and computer == "paper":
        return "player wins"
    
    return "computer wins"
```

```python
score = {"wins": 0,"losses": 0,"draws": 0,}

history = []
```

```python
print("Choose Difficulty:")
print("1. Easy")
print("2. Medium")
print("3. Hard")

difficulty = input("Enter your choice: ")
```

```python
if difficulty == "1":
    difficulty = "Easy"
elif difficulty == "2":
    difficulty = "Medium"
elif difficulty == "3":
    difficulty = "Hard"
else:
    difficulty = "Easy"

print("Difficulty:", difficulty)
print("Game starting...")
print()
```

```python
def play_round():
    player_choice = input("Enter your choice: ").lower()

    while player_choice not in choices:
        print("Invalid choice! Please enter rock, paper, or scissors.")
        player_choice = input("Enter your choice: ").lower()

    if difficulty == "Easy":
        computer_choice = random.choice(choices)

    elif difficulty == "Medium":
        if random.random() < 0.5:
            computer_choice = random.choice(choices)
        else:
            if player_choice == "rock":
                computer_choice = "paper"
            elif player_choice == "paper":
                computer_choice = "scissors"
            else:
                computer_choice = "rock"

    elif difficulty == "Hard":
        if player_choice == "rock":
            computer_choice = "paper"
        elif player_choice == "paper":
            computer_choice = "scissors"
        else:
            computer_choice = "rock"

    result = determine_winner(player_choice, computer_choice)

    print("Your choice:", player_choice)
    print("Computer choice:", computer_choice)
    print("Result:", result)

    if result == "player wins":
        score["wins"] += 1
    elif result == "computer wins":
        score["losses"] += 1
    else:
        score["draws"] += 1
    history.append({ "player": player_choice, "computer": computer_choice, "result": result})
```

```python
rounds = int(input("How many rounds do you want to play? "))
```

```python
for i in range(rounds):
    play_round()
```

```python
print("===== FINAL SCORE =====")
```

```python
print("Wins:", score["wins"])
```

```python
print("Losses:", score["losses"])
```

```python
print("Draws:", score["draws"])
```

```python
print("\n===== GAME HISTORY =====")

for i, game in enumerate(history, 1):
    print("Round", i,":", "You =", game["player"],"| Computer =", game["computer"],"| Result =", game["result"] )
```
